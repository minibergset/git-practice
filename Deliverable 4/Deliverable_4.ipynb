## Task 1: Temperatures
### 1. Initialization and validation
The class accepts C, F and K, including lowercase units. Invalid units and values below absolute zero raise ValueError.

```python
# Import the Temperature class from temperature.py.
from temperature import Temperature

# Create temperatures using lowercase and default units.
t = Temperature(25, 'c')
print(t)
print(Temperature(20))

# Try invalid values and print the errors.
for value, unit in [(0, 'X'), (-274, 'C'), (-460, 'F'), (-1, 'K')]:
    try:
        Temperature(value, unit)
    except ValueError as error:
        print(error)
```

### 2. Floating-point tolerance
Temperatures are compared in Kelvin using a tolerance of 1e-9.

```python
# Check if a tiny difference is treated as equal.
print(Temperature(0) == Temperature(0.0000000005))
```

### 3. Read-only attributes
The value and unit can be read, but cannot be changed directly.

```python
# Read the stored value and unit.
print(t.value, t.unit)
# Try to change a read-only attribute.
try:
    t.value = 30
except AttributeError:
    print('Value is read-only')
try:
    t.unit = 'F'
except AttributeError:
    print('Unit is read-only')
```

### 4. Conversion methods
The conversion methods return numbers. `convert_to()` returns a new Temperature object.

```python
# Convert the same temperature to each unit.
print(t.to_celsius())
print(t.to_fahrenheit())
print(t.to_kelvin())
# Create a new Temperature object in Fahrenheit.
print(t.convert_to('f'))
```

### 5. String representations
`str()` gives a readable temperature. `repr()` shows the object representation.

```python
# Show the two text formats.
print(str(t))
print(repr(t))
```

### 6. Comparisons
All six comparison operators work across different units. Unsupported types return NotImplemented from the comparison methods.

```python
# Create temperatures in different units.
a = Temperature(0, 'C')
b = Temperature(32, 'F')
c = Temperature(300, 'K')
# Try all six comparison operators.
print('==:', a == b)
print('!=:', a != c)
print('<:', a < c)
print('<=:', a <= b)
print('>:', c > a)
print('>=:', c >= b)
# Check what happens with an unsupported type.
print(a.__eq__(10))
```

### Testing – Task 1
`test_temperature.py` covers the six required test sections: initialization, validation, conversions (all five reference points), round-trip conversions, comparisons and formatting.

```python
# Import tools to run pytest from the notebook.
import sys
import subprocess

# Run the temperature tests.
result = subprocess.run(
    [sys.executable, "-m", "pytest", "-v", "test_temperature.py"],
    capture_output=True, text=True
)
# Show the results and check that all tests passed.
print(result.stdout)
assert result.returncode == 0
```

## Task 2: Fractions
### 1. Initializer
The numerator and denominator must be integers. The denominator must be positive.

```python
# Import the Fraction class from fraction.py.
from fraction import Fraction

# Create two fractions.
a = Fraction(1, 2)
b = Fraction(1, 4)
print(a)
# Check that a denominator of zero is rejected.
try:
    Fraction(1, 0)
except ValueError as error:
    print(error)
```

### 2. All comparisons
Fractions are compared using cross multiplication.

```python
# Compare the fractions using all six operators.
print('==:', a == Fraction(2, 4))
print('!=:', a != b)
print('<:', b < a)
print('<=:', b <= a)
print('>:', a > b)
print('>=:', a >= b)
```

### 3. Four other special methods
The four extra methods are `__str__`, `__repr__`, `__add__` and `__mul__`.

```python
# Show the two text formats.
print(str(a))
print(repr(a))
# Add and multiply the two fractions.
print(a + b)
print(a * b)
```

### Testing – Task 2
`test_fraction.py` tests initialization, all comparisons and the four extra methods.

```python
# Run the fraction tests.
result = subprocess.run(
    [sys.executable, "-m", "pytest", "-v", "test_fraction.py"],
    capture_output=True, text=True
)
# Show the results and check that all tests passed.
print(result.stdout)
assert result.returncode == 0
```
